### Beschreibung
Züri wie neu Meldungen: Gemeldet werden können sämtliche Schäden an der Infrastruktur der Stadt Zürich. Dabei kann es sich um ein Loch im Strassenbelag, ein Graffiti am Stadthaus oder eine durch Vandalen beschädigte Parkbank handeln. 

Zweck: «Züri wie neu» ist eine Online-Plattform, über die die Einwohnerinnen und Einwohner der Stadt Zürich auf Schäden an der städtischen Infrastruktur hinweisen können. «Züri wie neu» wird von der Stadtverwaltung moderiert und transparent geführt. Die Meldungen werden innerhalb von einem Arbeitstag den zuständigen Fachstellen zugewiesen und innert fünf Arbeitstagen abschliessend beantwortet. Fällt eine Meldung nicht in den Zuständigkeitsbereich der Stadtverwaltung, wird die Meldung anonymisiert der zuständigen dritten Stelle per E-Mail zugestellt.

## Quartiere Boundaries Data
Additionaly to the "ZüriWieNeu" Data. There is also data on the boundaries of the Quartiere of Zurich, which can be used to see differences among them and for mapping purpases.

Link: https://data.stadt-zuerich.ch/dataset/geo_statistische_quartiere

Again, there are multiple formates provided for the data. Formats of intrest for this project are:

- csv
- gpkg (GeoPackage, Projection CH1903+/LV95 or EPSG:2056)
- json (GeoJSON, Projection WGS84 or EPSG: 4326)
- shp (Shapefile, Projection CH1903+/LV95 or EPSG:2056)

In the Quartiere data ther are 3 seperat Layers:

DM_STATISTISCHE_QUARTIERE_B_P:

Beschriftungspunkte der statistischen Quartiere. Dieser Punktlayer beinhaltet Informationen zur optimalen Positionierung von Beschriftungen in Karten.

Attributes:
| Name | Definition |  |
|------|------------|--|
| KUERZEL | - |  |
| HALI | Horizontaler Versatz zum Ursprungspunkt; undefiniert = Center |  |
| OBJECTID | ESRI System-Identifier. Nicht stabil über Zeit. |  |
| OBJID | System-Identifier. Nicht stabil über Zeit. |  |
| VALI | Vertikaler Versatz zum Ursprungspunkt; undefiniert = Half |  |
| GEOMETRIE | Geometriefeld |  |
| NAME | Bezeichnung des statistischen Quartiers |  |
| ORI | Rotation der Beschriftung; undefiniert = 100.0; Masseinheit ist Gon. |  |


ADM_STATISTISCHE_QUARTIERE_V:

Dieser Layer enthält die exakten Grenzverläufe der Statistischen Quartiere und enthält die vollständige Attributtabelle. Dieser Layer eignet sich zur Berechnung von Flächenangaben oder bei räumlichen Verschnitten. Als kartographischer Layer ist er jedoch nicht geeignet.

Attributes:
| Name | Definition |  |
|------|------------|--|
| GEOMETRIE | Geometriefeld. |  |
| KNAME | Name des Stadtkreises. |  |
| OBJECTID | Interner System-Identifier. Nicht stabil über Zeit. |  |
| OBJID | System-Identifier. Nicht stabil über Zeit. |  |
| QNR | Offizielle ID des Statistischen Quartiers. |  |
| KNR | Offizielle ID des Stadtkreises. |  |
| QNAME | Name des Statistischen Quartiers. |  |


ADM_STATISTISCHE_QUARTIERE_MAP:

Dieser Layer kann für Datenvisualisierungen und kartographische Darstellungen verwendet werden. Er enthält die vollständige Attributtabelle und ist generalisiert, sprich: die Geometrie wurde vereinfacht und entspricht nicht der - aus Sicht der amtlichen Vermessung - korrekten Lage. 

Attributes:
| Name | Definition |  |
|------|------------|--|
| GEOMETRIE | Geometriefeld. |  |
| KNAME | Name des Stadtkreises. |  |
| KNR | Offizielle ID des Stadtkreises. |  |
| OBJECTID | Interner System-Identifier. Nicht stabil über Zeit. |  |
| OBJID | System-Identifier. Nicht stabil über Zeit. |  |
| QNAME | Name des Statistischen Quartiers. |  |
| QNR | Offizielle ID des Statistischen Quartiers. |  |
